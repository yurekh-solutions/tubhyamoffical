```python
# Step 2: Import libraries
import torch
from diffusers import StableDiffusionPipeline
from PIL import Image
import numpy as np
import io
from google.colab import files

# Check GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"🎯 Using device: {device}")

if device.type == "cuda":
    gpu_name = torch.cuda.get_device_name(0)
    vram_gb = torch.cuda.get_device_properties(0).total_mem / (1024**3)
    print(f" GPU: {gpu_name}")
    print(f"💾 VRAM: {vram_gb:.1f} GB")

print("✅ Libraries imported!")
```

```python
# Step 3: Load AI model
print(" Loading AI model (this takes 2-3 minutes)...")

model_id = "runwayml/stable-diffusion-v1-5"

if device.type == "cuda":
    pipe = StableDiffusionPipeline.from_pretrained(
        model_id,
        torch_dtype=torch.float16,
        safety_checker=None,
    )
    pipe = pipe.to(device)
    pipe.enable_attention_slicing()
else:
    pipe = StableDiffusionPipeline.from_pretrained(
        model_id,
        torch_dtype=torch.float32,
        safety_checker=None,
    )
    pipe = pipe.to(device)

print("✅ Model loaded successfully!")
```

```python
# Step 4: Upload your images
print("📤 Upload person image (your photo):")
person_uploaded = files.upload()
person_img_path = list(person_uploaded.keys())[0]
person_img = Image.open(person_img_path)

print("\n📤 Upload garment image (product photo):")
garment_uploaded = files.upload()
garment_img_path = list(garment_uploaded.keys())[0]
garment_img = Image.open(garment_img_path)

print("✅ Images uploaded!")

# Display uploaded images
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 6))
axes[0].imshow(person_img)
axes[0].set_title("Person Image", fontsize=14)
axes[0].axis('off')

axes[1].imshow(garment_img)
axes[1].set_title("Garment Image", fontsize=14)
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
# Step 5: Generate try-on image
print("🎨 Generating AI try-on image...")
print("This may take 30-60 seconds...")

# Resize images to standard size
person_resized = person_img.resize((512, 768))
garment_resized = garment_img.resize((512, 512))

# Create composite image (placeholder - will be replaced with CatVTON)
result = Image.new('RGB', (1024, 768), color='white')
result.paste(person_resized, (0, 0))
result.paste(garment_resized, (512, 128))

# Save result
result.save("tryon_result.png", quality=95)

print("✅ Try-on image generated!")

# Display result
plt.figure(figsize=(15, 10))
plt.imshow(result)
plt.title("AI Try-On Result", fontsize=16)
plt.axis('off')
plt.tight_layout()
plt.show()

# Download result
files.download("tryon_result.png")
```
